```python
# Cell 1: Imports
from pytube import YouTube
from moviepy.audio.io.AudioFileClip import AudioFileClip
import os
from pathlib import Path

```

```python
import os
from yt_dlp import YoutubeDL

def get_video_path(input_path, output_dir="downloads", filename="downloaded_video.mp4"):
    os.makedirs(output_dir, exist_ok=True)
    output_path = os.path.join(output_dir, filename)

    # Use yt_dlp to download video
    ydl_opts = {
        'format': 'mp4',
        'outtmpl': output_path,
        'quiet': False,
    }

    with YoutubeDL(ydl_opts) as ydl:
        ydl.download([input_path])

    return output_path
```

```python
# Cell 3: Extract audio from video
def extract_audio(video_path: str, audio_path: str = "audio.wav") -> str:
    clip = AudioFileClip(video_path)
    clip.write_audiofile(audio_path, codec='pcm_s16le')  # WAV format
    clip.close()
    print(f"✅ Audio extracted to: {audio_path}")
    return audio_path
```

```python
video = get_video_path("https://youtube.com/shorts/XcS7ih3bZzo?si=r8jogBC1lu1pqL2S")
audio = extract_audio(video)
```
